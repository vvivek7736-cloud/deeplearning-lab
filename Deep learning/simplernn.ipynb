```python
import tensorflow as tf
from tensorflow.keras import layers
from tensorflow.keras.datasets import imdb
from tensorflow.keras.utils import pad_sequences
```

```python
(x_train,y_train),(x_test,y_test)=imdb.load_data(num_words=10000)
```

```python
x_train=pad_sequences(x_train,maxlen=200)
x_test=pad_sequences(x_test,maxlen=200)
```

```python
model=tf.keras.Sequential([
    layers.Embedding(10000,32,mask_zero=True),
    layers.SimpleRNN(32),
    layers.Dense(1,activation='sigmoid')
])
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
hist=model.fit(x_train,y_train,epochs=5,batch_size=64,validation_split=0.1)
```

```python
l,a=model.evaluate(x_test,y_test)
a
```

```python

from tensorflow.keras.preprocessing.text import text_to_word_sequence

# Get word dictionary
word_index = imdb.get_word_index()

# Get review from user
review = input("Enter a movie review: ")

# Convert review into words
words = text_to_word_sequence(review)

# Convert words into numbers
sequence = [1]

for word in words:
    index = word_index.get(word, 2)

    if index != 2:
        index = index + 3

    if index >= 10000:
        index = 2

    sequence.append(index)

# Make the length 200
sequence = pad_sequences([sequence], maxlen=200)

# Predict sentiment
prediction = model.predict(sequence)

# Display result
if prediction[0][0] >= 0.5:
    print("Positive Review")
else:
    print("Negative Review")
```
