```python
#importing
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
df=pd.read_csv("polynomial.csv")
df.head(5)
```

```python
from sklearn.preprocessing import LabelEncoder
en=LabelEncoder()
df['position']=en.fit_transform(df['position'])
df.head(5)
```

```python
x=df[['level']]
y=df['salary']
```

```python
model=LinearRegression()
model.fit(x,y)
```

```python
plt.figure()
predicted=model.predict(x)
plt.scatter(x,y,color='red')
plt.plot(x,predicted)
plt.xlabel('level')
plt.ylabel('salary')
plt
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly=PolynomialFeatures(degree=2)
x_poly=poly.fit_transform(x)
model.fit(x_poly,y)
```

```python
plt.scatter(x,y,color='red')
plt.plot(x,model.predict(x_poly),color='blue')
plt.title('Polynomial Regression with degree 2')
plt.xlabel('Position level')
plt.ylabel('Salary')
plt.show()
```

```python
x=df[['level','position']]
y=df['salary']
```

```python
model=LinearRegression()
model.fit(x,y)
```

```python
import matplotlib.pyplot as plt
import numpy as np

a, b = np.meshgrid(x['level'], x['position'])
z = model.predict(np.c_[a.ravel(), b.ravel()]).reshape(a.shape)

ax = plt.axes(projection='3d')
ax.scatter(x['level'], x['position'], y)
ax.plot_surface(a, b, z, color='red', alpha=0.7)

plt.show()
```

```python
import matplotlib.pyplot as plt

ax = plt.axes(projection='3d')
ax.scatter(x['level'], x['position'], y)
ax.plot_trisurf(x['level'], x['position'], model.predict(x), color='red')

plt.show()
```
